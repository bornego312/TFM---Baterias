```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
csv_file = r"datos_baterias_mp\baterias_features_engineered.csv"
df = pd.read_csv(csv_file)
df_features = df.copy()

print(df_features.shape)
display(df_features.head())
```

### 1. Separación de varibales predictivas y objetivo

```python
target_columns = [
    "average_voltage",
    "max_delta_volume"
]

non_feature_columns = [
    "battery_formula",
    "working_ion",
    "average_voltage",
    "capacity_grav",
    "capacity_vol",
    "energy_grav",
    "energy_vol",
    "max_delta_volume",
    "max_voltage_step",
    "stability_charge",
    "stability_discharge",
    "framework_formula",
    "num_steps",
    "last_updated",
    "energy_estimada",
    "composition"
]
```

```python
feature_columns = [
    col
    for col in df_features.columns
    if col not in non_feature_columns
]
```

```python
X = df_features[feature_columns].copy()
```

```python
y_voltage = df_features["average_voltage"]

y_volume = df_features["max_delta_volume"]
```

### 2. Dividir dataset en train y test

```python
X_train, X_test, y_voltage_train, y_voltage_test, y_volume_train, y_volume_test = train_test_split(
    X,
    y_voltage,
    y_volume,
    test_size=0.2,
    random_state=42
)
```

### 3. Comprobación de dataset tras el split

```python
print("Dimensiones del conjunto de entrenamiento y prueba:")
print(f"X_train: {X_train.shape}")
print(f"X_test:  {X_test.shape}")

print(f"\ny_voltage_train: {y_voltage_train.shape}")
print(f"y_voltage_test:  {y_voltage_test.shape}")

print(f"\ny_volume_train: {y_volume_train.shape}")
print(f"y_volume_test:  {y_volume_test.shape}")
```

```python
print("Porcentaje de registros en cada conjunto:")

total = len(X_train) + len(X_test)

print(f"Train: {len(X_train)} registros ({len(X_train)/total*100:.2f}%)")
print(f"Test:  {len(X_test)} registros ({len(X_test)/total*100:.2f}%)")
```

```python
print("Valores nulos en cada conjunto:")

print("Nulos en X_train:", X_train.isna().sum().sum())
print("Nulos en X_test:", X_test.isna().sum().sum())

print("Nulos en y_voltage_train:", y_voltage_train.isna().sum())
print("Nulos en y_voltage_test:", y_voltage_test.isna().sum())

print("Nulos en y_volume_train:", y_volume_train.isna().sum())
print("Nulos en y_volume_test:", y_volume_test.isna().sum())
```

```python
print("Consistencia de dimensiones entre X e y:")

print(
    "Voltage train:",
    len(X_train) == len(y_voltage_train)
)

print(
    "Voltage test:",
    len(X_test) == len(y_voltage_test)
)

print(
    "Volume train:",
    len(X_train) == len(y_volume_train)
)

print(
    "Volume test:",
    len(X_test) == len(y_volume_test)
)
```

### 4. Guardado de datos

```python
X_train.to_csv("datos_baterias_mp/X_train.csv", index=False)
X_test.to_csv("datos_baterias_mp/X_test.csv", index=False)

y_voltage_train.to_csv("datos_baterias_mp/y_voltage_train.csv", index=False)
y_voltage_test.to_csv("datos_baterias_mp/y_voltage_test.csv", index=False)

y_volume_train.to_csv("datos_baterias_mp/y_volume_train.csv", index=False)
y_volume_test.to_csv("datos_baterias_mp/y_volume_test.csv", index=False)
```
